#### To create the user interface we will use the Python framework Streamlit

```python
#importando a biblioteca streamlit e atribuindo o apelido "st"
import streamlit as st  # type: ignore

# Library responsible for automatic page refresh
# # from streamlit_autorefresh import st_autorefresh  # removed dependency  # removed dependency
```

##### I chose to use Yahoo Finance for its flexibility and because it is free, making it an excellent option for building the prototype in my opinion.

```python
# Importing the Yahoo Finance library
import yfinance as yf
```

```python
# Importing libraries for date manipulation
from datetime import datetime, timedelta
```

```python
# Importing library to display interactive charts
import plotly.graph_objs as go
```

##### Creating the initial application script and running tests

```python
# List of available countries and time intervals
countries = ["Brazil", "United States"]
intervals = ["1d", "1wk", "1mo"]  # d - dia | wk - semana | mo - mês
```

```python
# Creating default filters for the search range (last 30 days)
start_date = datetime.now() - timedelta(days=30)
end_date = datetime.now()
```

```python
# Streamlit cached function to avoid redundant data requests
@st.cache_data()
def fetch_stock_data(stock, from_date, to_date, interval):
    """Fetches historical stock data from Yahoo Finance."""
    return yf.download(stock, start=from_date, end=to_date, interval=interval, progress=False)
```

```python
# Function to format a datetime object as a string
def format_date(dt, format="%Y-%m-%d"):
    """Converte datetime para string formatada."""
    return dt.strftime(format)
```

```python
# Function that creates a candlestick chart for OHLC data visualization
def plot_candlestick(df, ticker="UNKNOWN"):
    """Creates an interactive candlestick chart from OHLC data."""
    trace1 = {
        "x": df.index,
        "open": df["Open"],
        "close": df["Close"],
        "high": df["High"],
        "low": df["Low"],
        "type": "candlestick",
        "name": ticker,
        "showlegend": False,
    }

    data = [trace1]
    layout = go.Layout()

    return go.Figure(data=data, layout=layout)
```

```python
# Building the page sidebar with selection controls
sidebar_placeholder = st.sidebar.empty()
country_select = st.sidebar.selectbox("Select Country:", countries)
stocks = ["AAPL", "MSFT", "GOOGL"]  # List of stocks - customize with desired tickers
stock_select = st.sidebar.selectbox("Select stock:", stocks)
from_date = st.sidebar.date_input("Start Date:", start_date)
to_date = st.sidebar.date_input("End Date:", end_date)
interval_select = st.sidebar.selectbox("Select Interval:", intervals)
load_data = st.sidebar.checkbox("Load Data")

# Containers for dynamic chart rendering
chart_line = st.empty()
chart_candlestick = st.empty()
```

```python
# Defining initial page title and headers with smaller fonts
st.markdown("# Real-time Stock Chart Analysis")
st.markdown("## Stocks")
st.markdown("### Chart Analysis")
```

```python
# Refresh page every 5 seconds to allow real-time analysis
if st.sidebar.button("Refresh Data"):
    st.experimental_rerun()
count = 0

if count == 0:
    st.write("Count is zero")
elif count % 3 == 0 and count % 5 == 0:
    st.write("FizzBuzz")
elif count % 3 == 0:
    st.write("Fizz")
elif count % 5 == 0:
    st.write("Buzz")
else:
    st.write(f"Count: {count}")

# Validate date range and fetch/display stock data
if from_date > to_date:
    st.sidebar.error("Start Date cannot be later than End Date")
else:
    df = fetch_stock_data(stock_select, format_date(from_date), format_date(to_date), interval_select)
    try:
        # Display candlestick chart
        fig = plot_candlestick(df, stock_select)
        chart_candlestick.plotly_chart(fig)
        
        # Display closing price line chart
        chart_line.line_chart(df["Close"])
        
        # Optionally display raw data table
        if load_data:
            st.subheader("Dados")
            st.dataframe(df)
    except Exception as e:
        st.error(f"Error loading stock data: {e}")
```
