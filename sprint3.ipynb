#####  2. Controle de Taxa (Rate Limiting) e Cache
Prevent multiple requests in a short time:

```python
import time

# Function to limit requests over time
def rate_limit_check(last_request_time, min_interval=5):
    """Checks that the minimum interval between requests has been met."""
    current_time = time.time()
    if (current_time - last_request_time) < min_interval:
        raise ValueError("Requests are too frequent! Please wait a moment.")
    return current_time

# Global variable to store the time of the last request
last_request_time = time.time()

# Function to query stock data with rate limiting and cache
@st.cache_data(ttl=3600)  # Cache valid for 1 hour
def fetch_stock_data_with_limit(stock, from_date, to_date, interval):
    """Fetches historical data with rate limiting and cache."""
    global last_request_time
    last_request_time = rate_limit_check(last_request_time)
    return yf.download(
        stock, start=from_date, end=to_date, interval=interval, progress=False
    )
```
